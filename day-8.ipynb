# Data transformation using map()

```python
import pandas as pd
df = pd.read_excel("Excel_data.xlsx")
df
```

```python
# Query: double the salary of the employees

df6 = df.copy()
df6.salary = df6.salary.apply(lambda x : x * 2)
df6
```

```python
# Example-3: Apply() function
# Query: Create a new column which is having Salary/age
# motive: use of apply() function on complete dataframe and access multiple columns.

df6["salary_per_year_age"] = df6.apply(lambda row : row['salary'] / row['Age'], axis = 1)
df6
```

# map()

```python
# def: map() is used for element_wise transformation.

# example-1: 
df7 = df.copy()

# Query: convert all the strings in upper case
df7.Name = df7.Name.map(lambda val : str(val).upper())
df7
```

```python
# Example-2:

# Query: Convert professions from english language to telugu.
df.Profession.unique()
```

```python
profession_map = {
    'Teacher' : 'Saarr',
    '      Boxer' : 'peddi',
    'Scientist' : 'sastravedu',
    'Chef' : 'vantavadu',
    'Doctor' : 'Vaidyudu',
    'Artist' : 'kalakar',
    'Comedian' : 'Haasya',
    'Software Engineer' : 'kulodu',
    'Detective' : 'gudachariya',
    'Pilot' : 'Galodu'
}

df7['Prof_telugu'] = df7.Profession.map(profession_map)
df7
```

# Grouping function

```python
# groupby(): group the data based upon category

df.groupby('Profession').Name.count()
```

```python
# Query: find the average salary based upon profession
df.groupby('Profession').salary.mean()
```

```python
# find average salary based upon age
df.groupby('Age').salary.mean()
```

```python
# agg(): can apply multiple aggregate functions together on data

df[["Age","salary"]].agg(["mean", "sum"])
```

```python
# apply different aggregate functions on different columns
df.agg({
    "Age" : ["min", "max"],
    "salary" : ["mean", "min"]
})
```

```python
# groupby() + agg()
df.groupby("Profession").agg({
    "Age" : ["min", "max"],
    "salary" : ["mean", "min"],
})    
```
