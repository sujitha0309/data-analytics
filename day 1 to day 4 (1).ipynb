##### creating a new column in the dataframe

```python
# Here, the rank function is a pandas function
# The rank function gives rank 1 to the minimum salary and a higher rank value to higher salaries

df['rank123'] = df['salary'].rank(ascending = False)
df.sort_values('rank')

# sort_values() sorts the dataframe based on a column value.
```

# Data filtering and indexing (loc,iloc,where())

```python
# query: filter all the records  where salary > 100000
df.salary > 100000
```

```python
df[df.salary > 100000]
```

```python
# Query : filter all the records where salary > 100000 & salary < 150000

df[df.salary > 100000 & 150000]
```

```python
df[(df.salary > 100000) & (df.salary < 150000)]
```

```python
# Query : filter all the records where salary < 100000 | or salary > 150000

df[(df.salary > 100000) | (df.salary > 150000)]
```
