```python
# Query : Find the person record who is getting highest salary.
# hint : max()
import pandas as pd
df = pd.read_excel("Excel_data.xlsx")
df
```

```python
# Query : Find the person record who is getting highest salary.
max(df.salary)
```

```python
df[df.salary == df.salary.max()] 
```

```python
# Query: Find the person record who is getting lowest salary.
# Hint: min()

min(df.salary)
```

```python
df[df.salary == df.salary.min()]
```

```python
df.salary == df.salary.min()
```

### where()

```python
### where() : used for filtering the data
# where() function returns NaN in place of those values
# which don't statisfied the condition

df1 = df.copy()
df1 = df1.Age.where(df1.Age > 50,1)
df1
```

```python
# if you dont want NaN records
df[df.Age > 50]
```

# indexing: loc & iloc

```python
# loc & iloc is used for filtering
# these helps to select particular row and column in the dataframe
```

```python
# Query: print records 1 to 4 and columns id, name, age only.
df.loc[1:4, 'ID': 'Age'] # df.loc[start_row : end_row, col_start: col_end]
```

```python
# Query: print records 10 to 40, and columns Name:profession only.

df.loc[10:40, "Name" : "Profession"]
```
