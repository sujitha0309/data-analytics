### 5. Handling of missing values(NaN/Null)

```python
# finding the null values in dataframe:
import pandas as pd
```

```python
df = pd.read_excel("Excel_data.xlsx")
df
```

```python
df.isnull().sum()
```

```python
# check which column have how many null values.
df.isnull()
```

```python
# records with null values
df[df.Age.isnull()]
```

```python
df.Age.isnull()
```

```python
df.Age
```

```python
df1 = df.copy()

# delete the recordds which is having null values
# how = 'any' : if even a single value is null then delete the column.
# how = "all" : delete column only if all values in the column is null.
df1 = df1.dropna()
df1
```

```python
df1.shape
```

```python
df2 = df.copy()

# delete the columns with null value.
# how = 'any' : if even a single value is null then delete the column.
# how = "all" : delete column only if all values in the column is null
df2 = df2.dropna(axis = 1)
df2
```

```python
# replacing the null values with some other value

df3 = df.copy()
df3.Age = df3.Age.fillna(23)
df3.isnull().sum()
```

# Finding & Handling duplicate values

```python
# 2nd and more duplicated
# complete record is matching with some other record.
df[df.duplicated()]
```

```python
# showing all the duplicate records.
df[df.duplicated(keep = False)]
```

```python
# finding duplicates based upon specific columns.
#Query: count all the records where age & salary is matching/duplicate
df [df.duplicated(keep = False, subset = ['Name', 'Profession', 'salary'])].count()
```

```python
# delete specific column based duplicate records
df5 = df.copy()
df5 = df5.drop_duplicates(subset = ['Age', 'salary'])
df5.shape
```
