```python
import pandas as pd
df = pd.read_excel("Excel_data.xlsx")
df
```

```python
df.iloc[0:20, 0:5]
```

```python
# iloc using step size
df.iloc[0:200:10, 0:5:2]
```

```python
df.columns
```

```python
# iloc showing any random columns
df.iloc[0:200:10, [1,4]]
```

```python
df.iloc[1:10, [1,3,4]]
```

```python
# iloc to get specific value
print(df.iloc[1,5])
```

```python
# loc to get age and salary columns
df.loc[0:5, ["Age", "salary"]]
```

```python
# Comparision condition inside the loc
df.Age > 50
```

```python
# comparision condition inside the loc

df.loc[df.Age > 50, ["Age", "salary"]]
```

```python
# salary of people Age > 50 & Age < 70
df.loc[(df.Age > 50) & (df.Age < 70) ,["Age", "salary"]]
```

```python
# iloc don't works with pandas series
# .values will convert it into numpy array
df.iloc[(df.Age > 50).values, 0:4]
```

```python
df.iloc[(df.Age > 50), 0:4]
```

```python
type((df.Age > 50).values)
```

### updating a value inside the dataframe

```python
# replace/update all the age > 50 with 18
df[df.Age > 50]
```

```python
df[df.Age > 50].Age
```

```python
df.loc[df.Age > 50, "Age"] = 18
```

```python
df
```

```python
### at & iat for single value access
# at and iat is faster than loc and iloc

df.at[0, "Name"] = "Mahesh babu"
df
```

```python
df.iat[0,1] = "vijay devarkonda"
df
```
